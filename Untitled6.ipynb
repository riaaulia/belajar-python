```python
def jumlah(angka):
  return sum(angka)
def rata_rata(daftar):
  return jumlah(daftar) / len(daftar)
nilai = [90, 80, 75, 85]
print(rata_rata(nilai))
print(rata_rata([90, 95]))
```

```python
data = [
    {"nama": "ani", "nilai": 80},
    {"nama": "budi", "nilai": 65},
    {"nama": "cici", "nilai": 90},
]
#dengan loop, tampilkan nama tiap orang beserta nilainya
for x in data:
  print(x["nama"], x["nilai"])
#pakai if untuk menampilkan nilai >= 70
for x in data:
  if x["nilai"] >= 70:
    print(x["nama"], x["nilai"])
#hitung rata-rata
def jumlah(angka):
  return sum(angka)
def rata_rata(daftar):
  return jumlah(daftar) / len(daftar)
nilai = []
for x in data:
  nilai.append(x["nilai"])
print(rata_rata(nilai))
```
